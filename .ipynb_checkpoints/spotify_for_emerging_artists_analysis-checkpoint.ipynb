## Overview

This analysis evaluates how different release formats (Singles vs. Albums) influence artist and track popularity to create a cohesive data-driven product strategy recommendation for emerging artists using 16 years of Spotify data. 

## Insight 1: Singles on the rise...but not albums?

#### Market Insights
- The Pivot: Release trends confirm that consumers of Spotify favor single releases over albums. This shift minimizes the commercial risk of a single project underperforming and maximizes consistent algorithmic discovery over time.
- Algorithmic Favoritism: Spotify’s recommendation system rewards frequent, consistent content. Releasing individual tracks is less resource-intensive than album releases and keeps artists active in the algorithm.
- Resource Realignment: Record labels and independent distributors should prioritize continuous funding structures for serialized singles or rolling EPs rather than sinking long-term budgets into highly concentrated, singular album rollouts.

```python
spotify_df_single_bartchart = sns.countplot(data=spotify_df_single, x="release_year", color="#dd8452")
ticks = spotify_df_single_bartchart.get_xticks()
labels = [label.get_text() for label in spotify_df_single_bartchart.get_xticklabels()]
step = 4
sns.set_theme(style="whitegrid")
spotify_df_single_bartchart.set_xticks(ticks[::step])
spotify_df_single_bartchart.set_xticklabels(labels[::step], rotation=0)
plt.title("Growth of Single Releases (2009-2025)", fontsize=14, fontweight='bold', pad=15)
plt.yticks(np.arange(0, 450, 50))
y_counts = [p.get_height() for p in spotify_df_single_bartchart.patches]
x_indices = np.arange(len(y_counts))
degree = 3  
coefficients = np.polyfit(x_indices, y_counts, degree)
trend_curve = np.polyval(coefficients, x_indices)
plt.plot(ticks, trend_curve, color='red', linewidth=2, linestyle='--')
for container in spotify_df_single_bartchart.containers:
    spotify_df_single_bartchart.bar_label(container, fmt='%.0f', padding=1, fontsize=8, zorder=5)
```

```python
spotify_df_album_bartchart = sns.countplot(data=spotify_df_album, x="release_year", color="#4c72b0")
ticks = spotify_df_album_bartchart.get_xticks()
labels = [label.get_text() for label in spotify_df_album_bartchart.get_xticklabels()]
step = 4
sns.set_theme(style="whitegrid")
spotify_df_album_bartchart.set_xticks(ticks[::step])
spotify_df_album_bartchart.set_xticklabels(labels[::step], rotation=0)
plt.title("Growth of Album Releases (2009-2025)", fontsize=14, fontweight='bold', pad=15)
plt.yticks(np.arange(0, 400, 50))
y_counts = [p.get_height() for p in spotify_df_album_bartchart.patches]
x_indices = np.arange(len(y_counts))
degree = 3  
coefficients = np.polyfit(x_indices, y_counts, degree)
trend_curve = np.polyval(coefficients, x_indices)
plt.plot(ticks, trend_curve, color='red', linewidth=2, linestyle='--')
for container in spotify_df_album_bartchart.containers:
    spotify_df_album_bartchart.bar_label(container, fmt='%.0f', padding=1, fontsize=8, zorder=5)
```

## Insight 2: Emerging or less popular artists gain more traction from singles

```python
bins = [0, 40, 70, 100]
labels_list = ['Low Popularity (0-40)', 'Medium Popularity (40-70)', 'High Popularity (70-100)']
plot_data = spotify_df[spotify_df['album_type'] != 'compilation'].copy()
plot_data['artist_popularity'] = pd.to_numeric(plot_data['artist_popularity'], errors='coerce')
plot_data['track_popularity'] = pd.to_numeric(plot_data['track_popularity'], errors='coerce')
plot_data['artist_popularity'] = pd.cut(
    plot_data['artist_popularity'], 
    bins=bins, 
    labels=labels_list, 
    include_lowest=True)
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")
tier_order = ['High Popularity (70-100)', 'Medium Popularity (40-70)', 'Low Popularity (0-40)']
popularity_rank_groups_plot = sns.barplot(
    data=plot_data,
    x='artist_popularity',
    y='track_popularity',
    hue='album_type',
    order=tier_order,  # FIX 2: Explicitly use tier_order to avoid global variable issues
     palette={'album': '#4c72b0', 'single': '#dd8452'},
    errorbar=None)
plt.title('Track Performance: Singles vs Albums by Popularity Ranking', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Artist Popularity Ranking Groups', fontsize=12, labelpad=10)
plt.ylabel('Average Track Popularity', fontsize=12, labelpad=10)
plt.legend(title='Format Type', loc='upper right')
for container in popularity_rank_groups_plot.containers:
    popularity_rank_groups_plot.bar_label(container, fmt='%.0f', padding=5, fontsize=10)
plt.ylim(0, 70)
```

#### Background
- Spotify ranks tracks and artists on a 0 to 100 "Popularity Score" with higher numbers indicate greater overall popularity and current streaming velocity.
- The popularity ranking is not a linear scale; it is exponential. A jump from 50 to 70 represents a massive increase in actual listenership.
- The calculation relies heavily on plays over the last 30 days, listener engagement ratios (saves and playlist additions), and low skip rates.

#### Takeaways
1. Emerging Artists (Low Popularity: 0–40)
    - Metric: Singles vastly outperform albums with an average score of 28.0 vs. 23.1.
    - Analysis: For newer or lesser-known acts, full album releases underperform. Singles act as low-friction entry points that are more easily picked up by algorithmic playlist structures, expanding baseline reach for small names.

2. Medium-Tier Artists (Medium Popularity: 40–70)
    - Metric: Performance between release formats is functionally identical (45.4 for albums vs. 45.3 for singles).
    - Analysis: Artists at this level maintain a stable baseline audience. Neither format provides an inherent algorithmic or performance advantage over the other. Strategy at this tier can be dictated by creative preference rather than structural optimization.

3. Mainstream Artists (High Popularity: 70–100)
    - Metric: Comprehensive album formats maximize average track scores, yielding top-tier performance on the platform.
    - Analysis: Established superstars possess dedicated, active fanbases that consume full-length projects on release day. This concentrated listening velocity amplifies the scores of deeper album tracks, a lift that individual single releases do not uniquely replicate.

## Insight 5: Likelihood of an Emerging Artist Making it to Mainstream 

#### What is the probability that an emerging artist, whose prior track peaked between 20 and 58, will produce a track that breaks into the top 58 to 100?
- Simulation: Monte Carlo
- Number of simulations: 100,000 hypothetical pairings of tracks and artists
- Data source: Spotify (2009 to 2025)
- Where I downloaded it: Kaggle

```python
# Identify the pool of "Emerging Artists" (who have at least one track that is ranked between 20 and 58)
emerging_artists_pool = spotify_df[
    (spotify_df["track_popularity"] >= 20)
    & (spotify_df["track_popularity"] <= 58)]["artist_name"].unique()

# Filter the dataframe to only include the full catalog of these specific artists
emerging_artists_df = spotify_df[
    spotify_df["artist_name"].isin(emerging_artists_pool)]

# Run a Monte Carlo simulation
# We randomly sample an artist from the pool, then sample another track from their catalog
n_simulations = 100000
success_count = 0
success_array = np.zeros(n_simulations) 

np.random.seed(42)  # The Ultimate Question of Life, the Universe, and Everything, is 42

for i in range(n_simulations):
    # Pick a random artist from our emerging pool
    random_artist = np.random.choice(emerging_artists_pool)
    # Get all tracks belonging to this artist
    artist_tracks = emerging_artists_df[emerging_artists_df["artist_name"] == random_artist]["track_popularity"].values
    if len(artist_tracks) > 1:
        # Simulate producing a new/different track from their catalog
        simulated_track_popularity = np.random.choice(artist_tracks)
        success_array[i] = simulated_track_popularity
        if simulated_track_popularity > 58:
            success_count += 1
            success_array

# Calculate simulation probability
simulation_probability = (success_count / n_simulations) * 100

# Calculate empirical baseline
total_tracks = len(spotify_df[spotify_df["track_popularity"] > 0])
breakthrough_tracks = len(spotify_df[spotify_df["track_popularity"] > 58])

# Calculate p-value of result
p_null = breakthrough_tracks / total_tracks
standard_error = np.sqrt((p_null * (1 - p_null)) / n_simulations)
z_score = ((simulation_probability / 100) - p_null) / standard_error
p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))
significance_text = " (Result is statistically significant)" if p_value < 0.05 else " (Result is not statistically significant)"
disadvantage_multiplier = (p_null * 100) / simulation_probability
```

```python
all_emerging_artists = spotify_df[(spotify_df["track_popularity"] <= 58)]["artist_name"].unique()
all_emerging_artists_ct = len(all_emerging_artists)
emerging_out_of_all = len(emerging_artists_pool) / all_emerging_artists_ct
emerging_out_of_all
```

```python
print("\n" + "=" * 55)
print("   MONTE CARLO SIMULATION:  BREAKING INTO MAINSTREAM")
print("=" * 55 + "\n")
print(f"👤 Total Emerging Artists in Pool: {all_emerging_artists_ct}")
print(f"🎼 Total Emerging Artists with a Track Ranked 20-58: {len(emerging_artists_pool)} ({emerging_out_of_all:.2f}%)")
print(f"🎲 Simulated Runs: {n_simulations:,}")
print(f"📊 Empirical Baseline (all tracks above rank 58): {p_null * 100:.2f}%")
print("")
print(f"📌 Predictive Probability of a track breaking through rank 58: \033[1m{simulation_probability:.1f}%\033[0m")
print(f"⚠️ Emerging artists face a significant systemic bottleneck: \033[1m{disadvantage_multiplier:.1f}x\033[0m LESS likely")
print("\n" + "=" * 55)
```

# CONCLUSION

## Executive Summary
This report evaluates market trends, track performance, and algorithmic distribution of content over the last 16 years of Spotify to provide strategic recommendations for promoting emerging artists and ecosystem diversification. This analysis reveals a highly concentrated ecosystem where emerging talents face extreme systemic barriers to entry into mainstream. By analyzing specific trends regarding type (singles vs. albums) performance and who dominate the Spotify algorithm, the follow are recommendations to ensure long-term platform health, improve growth for emerging artists, and optimize discovery mechanics.

## EDA and Background
#### Problem Statement
The current Spotify ecosystem is highly concentrated. Emerging talents face systemic barriers to entry into the mainstream catalog, driven by algorithm mechanics that heavily favor established, high-performing artists.

#### Market Trends (2009 to 2025)
The trend of releasing music as standalone singles has risen exponentially with no sign of slowing down. In 2025 alone, 121 (+137%) more tracks were released as singles than albums.

## Key Findings



#### The Mainstream Threshold
An artist whose prior track peaked between a popularity rank of 20 and 58 faces a narrow 12.9% empirical probability of breaking through the rank 58 threshold into the mainstream catalog.

#### Algorithmic Dominance
The top 15 established global artists command 30% share of high-ranking tracks (ranked between 70 to 100). The current recommendation algorithm heavily favors already established well-known artists (Taylor Swift, Bad Bunny, The Weeknd), slimming visibility for emerging talent.

#### Strategic Release of Singles
Spotify’s recommendation system rewards frequent, consistent content. Releasing individual tracks is less resource-intensive than album releases and keeps artists active in the algorithm. For emerging artists, releasing standalone singles yields higher track ranking compared to full-length album drops. The inverse is true for highly popular artists who benefit more from album drops. Singles are great low-barrier entry points for emerging artists on Spotify, hence their exponential growth over the years.

#### The Market Split
After eliminating all tracks that rank 0 from the dataset, 51% of tracks successfully reside in the mainstream tier.

#### Emerging Artists Face a Massive Disadvantage
The 100,000 simulations prove that an emerging artist who has produced a track ranked 20-58 in popularity is 4 times less likely (51% / 12.9%) to produce a mainstream track than the dataset's empirical baseline.

## Implications
Out of the pool of emerging artists, 75% released a track that ranks between 20 to 58. However, only 13% of them will break into mainstream and release a track that ranks above the 58 threshold. The difficulty rising stars face is not an isolated individual issue, but rather a structural flaw within the Spotify's recommendation and discovery system. When recommendation feeds prioritize historical engagement metrics over content variety, it keeps superstar tracks entrenched at the top while gatekeeping new entries from emerging talent.

#### Risks for the Spotify's Stability
1. Catalog Stagnation
    - Relying on a small pool of mega-hits leads to cultural stagnation and listening fatigue.

2. Loss of Trust
    - If emerging  creators realize the platform's infrastructure is against them, they are likely to move their efforts to alternative distribution channels.

3. Increased Vulnerability
    - Continued dependence on a highly concentrated pool of royalty users reduces Spotify's bargaining leverage during major record label negotiations.

## Recommendations (For Spotify)
1. Discovery Baseline
    - Introduce an internal structural constraint within personalized frameworks like Discover Weekly and the home feed that guarantees a fixed footprint for emerging artists. 

2. A Framework that Automates Planning
    - Create an in-app calculator that automates artists' planning and release cadence using the artist's data within the platform's infrastructure.
    - The calculator pulls the artist's historial data and listener base profiles.
    - Metrics: algorithmic streams, added to playlists, searches, save rate, completion rate, drop-off point, follower conversion.
    - Informs the exact holiday/season for emerging talent to avoid releasing in so the track  doesn't compete with major labels. 
    
3. Recommender System Optimization (RSO)
    - Provide transparency for artists who need to understand how close they are to triggering an algorithmic push through an internal tool. Create an in-app Recommender System Optimization (RSO) tool allows artists a visual interface that shows where their track sits relative to others.
    - For example, Spotify can provide powerful insights such as: "Your track is at 14% save rate, you need X more saves to be pushed into Discover Weekly".

## Importance
- Empower emerging talent within a highly competitive industry to optimize their content distribution in Spotify. Helping emerging artists break through using data and algorithmic tools rather than relying on luck transforms the platform from just a streaming service into an active engine for creator growth.
- Spotify automates the burden of independent artists who do data analytics, tracking, and promotion on their own, allowing more time for creating music and touring.
